```python
from langchain_openai import ChatOpenAI
from langchain_core.tools import tool
from langchain_core.messages import HumanMessage
import requests
```

```python
# Tool create
```

```python
@tool
def multiply(a:int,b:int)->int:
  """Given two numbers a and b this tool returns their product"""
  return a*b
```

```python
print(multiply.invoke({'a':4,'b':5}))
```

```python
# tool binding
```

```python
from getpass import getpass
OPENAI_API_KEY=getpass("Enter OPENAI API key")
```

```python
import os
os.environ["OPENAI_API_KEY"]=OPENAI_API_KEY
```

```python
llm=ChatOpenAI(model="gpt-4o-mini")
```

```python
llm_with_tools=llm.bind_tools([multiply])
```

```python
llm_with_tools.invoke("Hello how are you")
```

```python
llm_with_tools.invoke('Can u multiply 5 and 8')
```

```python
# Just observe the above two responses.
# For 1st content is there and for 2nd response content is empty
# Additionally you can see tool_calls in the second response
```

```python
# Note: LLM does not actually run the tool- it just suggestes the tool and the input arguments.
# Actual execution is handled by langchain or programmer
```

# Tool Execution

```python
# Step where actual python function is run using the input arguments that the LLM suggested during tool calling
```

```python
result=llm_with_tools.invoke('Can u multiply 5 and 8')
```

```python
result
```

```python
result.tool_calls
```

```python
result.tool_calls[0]
```

```python
multiply.invoke(result.tool_calls[0])
```

```python
# Tell llm the tool call is returning message .
# For that we need to maintain the conversation history
```

```python
query=HumanMessage("can u multiply 5 and 9")
```

```python
messages=[query]
```

```python
result=llm_with_tools.invoke(messages)
```

```python
messages.append(result)
```

```python
tool_result=multiply.invoke(result.tool_calls[0])
```

```python
messages.append(tool_result)
```

```python
messages
```

```python
final_result=llm_with_tools.invoke(messages)
```
