```python
# Method 2 of creating custom tool (Using Structured Tool class and pydantic)
```

```python
# Structured tool in langchain is a special type of tool where the input to the tool follows a structured schema,typically defined using a Pydantic model
```

```python
from langchain_core.tools import StructuredTool
from pydantic import BaseModel,Field
```

```python
class MultiplyInput(BaseModel):
  a:int=Field(required=True,decription="The first number to multiply")
  b:int=Field(required=True,description="The second number to multiply")
```

```python
def multiply_func(a:int,b:int)->int:
  return a*b
```

```python
multiply_tool=StructuredTool.from_function(
    func=multiply_func,
    name="Multiply",
    description="Multiply two numbers",
    args_schema=MultiplyInput
)
```

```python
result=multiply_tool.invoke({'a':3,'b':5})
```

```python
result
```

```python
print(multiply_tool.description)
print(multiply_tool.name)
print(multiply_tool.args)
```

```python
# Step 3 of creating custom tool - (Using BaseTool class)
```

```python
# Base tool is the abstract base class for all tools in Langchain.
# It defines the core structure and interface that any tool must follow,
# Whether its simple one-liner or a fully customized function
```

```python
# All other tool type like @tool , StructuredTool are built on top of BaseTool
```

```python
# BaseTool class is the class which tells how tool behaves in langchain
```

```python
from langchain_core.tools import BaseTool
from typing import Type
```

```python
# arg scheme using pydantic
class MultiplyInput(BaseModel):
  a:int=Field(required=True,decription="The first number to multiply")
  b:int=Field(required=True,description="The second number to multiply")
```

```python
class MultiplyTool(BaseTool):
  name:str="multiply"
  description:str="Multiply two numbers"
  args_schema:Type[BaseModel]=MultiplyInput

  def _run(self,a:int,b:int)->int:
    return a*b

```

```python
multiply_tool=MultiplyTool()
```

```python
result=multiply_tool.invoke({'a':4,'b':8})
```

```python
result
```

```python
print(multiply_tool.description)
print(multiply_tool.name)
print(multiply_tool.args)
```
